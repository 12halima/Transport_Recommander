```python
import os
from pyrosm import OSM
import pandas as pd

folder_path = r"C:\Users\hp\Documents\OSM_SP"

# Lister les fichiers .osm.pbf
files = [f for f in os.listdir(folder_path) if f.endswith(".osm.pbf")]
print(f"Nombre de fichiers trouvés : {len(files)}")
print("Exemples :", files[:10])

def inspect_osm(file_path, n=5):
    print(f"\n🔹 Lecture du fichier : {os.path.basename(file_path)}")
    osm = OSM(file_path)
    
    # Extraire les routes
    try:
        roads = osm.get_network(network_type="all")  # <-- méthode actuelle
        print(f"Nombre de routes trouvées : {len(roads)}")
        print(f"Aperçu des {n} premières lignes :")
        display(roads.head(n))
    except Exception as e:
        print("Erreur lors de l'extraction des routes :", e)
    
    # Extraire les points
    try:
        points = osm.get_pois()  # <-- points d'intérêt
        print(f"Nombre de POIs trouvés : {len(points)}")
        display(points.head(n))
    except Exception as e:
        print("Erreur lors de l'extraction des POIs :", e)

# Tester sur les 2-3 premiers fichiers
for file in files[:3]:
    inspect_osm(os.path.join(folder_path, file))
```

```python
import os
import pandas as pd
from pyrosm import OSM

# 📁 Dossier contenant les fichiers PBF
FOLDER_PATH = r"C:\Users\hp\Documents\OSM_SP"

# ✅ Colonnes attendues
selected_columns = [
    "id", "highway", "junction", "lanes", "maxspeed", "oneway",
    "busway", "psv", "motorcar", "access", "width", "smoothness",
    "name", "geometry", "length"
]

results = []

files = [f for f in os.listdir(FOLDER_PATH) if f.endswith(".osm.pbf")]

print(f"📂 Nombre total de fichiers : {len(files)}\n")

for file in files:
    file_path = os.path.join(FOLDER_PATH, file)
    print(f"🔍 Analyse : {file}")

    try:
        osm = OSM(file_path)
        roads = osm.get_network(network_type="driving")

        if roads is None or roads.empty:
            results.append({
                "file": file,
                "status": "❌ Aucun réseau routier",
                "missing_columns": selected_columns
            })
            print("   ❌ Aucun réseau trouvé\n")
            continue

        existing_cols = roads.columns.tolist()
        missing_cols = [c for c in selected_columns if c not in existing_cols]

        results.append({
            "file": file,
            "status": "✅ OK" if not missing_cols else "⚠️ Colonnes manquantes",
            "missing_columns": missing_cols
        })

        if missing_cols:
            print(f"   ⚠️ Colonnes manquantes : {missing_cols}\n")
        else:
            print("   ✅ Toutes les colonnes présentes\n")

    except Exception as e:
        results.append({
            "file": file,
            "status": "❌ Erreur de lecture",
            "missing_columns": selected_columns
        })
        print(f"   ❌ Erreur : {e}\n")

# 📊 Résumé global
df_results = pd.DataFrame(results)

print("\n📊 RÉSUMÉ GLOBAL")
print(df_results["status"].value_counts())

df_results
```

```python
import os
import gc
from pyrosm import OSM

FOLDER_PATH = r"C:\Users\hp\Documents\OSM_SP"
OUTPUT_FOLDER = os.path.join(FOLDER_PATH, "cleaned_files")
os.makedirs(OUTPUT_FOLDER, exist_ok=True)

selected_columns = [
    "id", "highway", "junction", "lanes", "maxspeed", "oneway",
    "busway", "psv", "motorcar", "access", "width", "smoothness",
    "name", "geometry", "length"
]

osm_files = [f for f in os.listdir(FOLDER_PATH) if f.endswith(".osm.pbf")]

print(f"📂 Nombre total de fichiers : {len(osm_files)}")

# 🚦 Types de routes utiles (réduction mémoire énorme)
ALLOWED_HIGHWAYS = [
    "motorway", "trunk", "primary", "secondary",
    "tertiary", "residential", "busway"
]

for i, filename in enumerate(osm_files, 1):
    print(f"\n[{i}/{len(osm_files)}] 🔹 {filename}")

    try:
        osm_path = os.path.join(FOLDER_PATH, filename)

        # ⛔ Skip fichiers trop gros (>300MB)
        if os.path.getsize(osm_path) > 300 * 1024 * 1024:
            print("   ⏭️ Fichier trop volumineux → ignoré")
            continue

        osm = OSM(osm_path)

        gdf = osm.get_network(
            network_type="driving",
            extra_attributes=selected_columns
        )

        if gdf is None or gdf.empty:
            print("   ⚠️ Aucune donnée")
            continue

        # 🚦 Filtrer routes utiles
        gdf = gdf[gdf["highway"].isin(ALLOWED_HIGHWAYS)]

        if gdf.empty:
            print("   ⚠️ Routes filtrées vides")
            continue

        # ➕ Ajouter colonnes manquantes
        for col in selected_columns:
            if col not in gdf.columns:
                gdf[col] = None

        gdf = gdf[selected_columns]

        output_path = os.path.join(
            OUTPUT_FOLDER,
            filename.replace(".osm.pbf", ".gpkg")
        )

        gdf.to_file(output_path, driver="GPKG")
        print("   ✅ Sauvegardé")

        # 🧹 Nettoyage mémoire CRUCIAL
        del gdf, osm
        gc.collect()

    except Exception as e:
        print(f"   ❌ Erreur ignorée : {e}")
        gc.collect()
```

```python
import os
import geopandas as gpd
import gc

# 🗂️ Chemins
FOLDER_PATH = r"C:\Users\hp\Documents\OSM_SP\cleaned_files"  # dossier contenant les GPKG filtrés
OUTPUT_FILE = r"C:\Users\hp\Documents\OSM_SP\OSM_Spain_Fused.gpkg"

# Créer le dossier de sortie si non existant
os.makedirs(os.path.dirname(OUTPUT_FILE), exist_ok=True)

# ✅ Colonnes à conserver
selected_columns = [
    "id", "highway", "junction", "lanes", "maxspeed", "oneway",
    "busway", "psv", "motorcar", "access", "width",
    "smoothness", "name", "geometry", "length"
]

# Lister tous les fichiers GPKG
gpkg_files = [f for f in os.listdir(FOLDER_PATH) if f.endswith(".gpkg")]
print(f"📂 Nombre de fichiers à fusionner : {len(gpkg_files)}")

# Liste pour stocker les GeoDataFrames
gdf_list = []

for i, file_name in enumerate(gpkg_files, start=1):
    file_path = os.path.join(FOLDER_PATH, file_name)
    print(f"[{i}/{len(gpkg_files)}] 🔹 Lecture : {file_name}")
    
    try:
        gdf = gpd.read_file(file_path)
        
        # garder uniquement les colonnes sélectionnées si elles existent
        cols_to_keep = [c for c in selected_columns if c in gdf.columns]
        gdf = gdf[cols_to_keep]
        
        gdf_list.append(gdf)
        
        # Libérer la mémoire
        del gdf
        gc.collect()
        
    except Exception as e:
        print(f"❌ Erreur lecture {file_name} : {e}")

# Fusionner tous les GeoDataFrames
print("\n🔗 Fusion des fichiers...")
try:
    fused_gdf = gpd.GeoDataFrame(pd.concat(gdf_list, ignore_index=True), crs=gdf_list[0].crs)
    print(f"✅ Fusion terminée, total lignes : {len(fused_gdf)}")
    
    # Sauvegarder en GPKG
    fused_gdf.to_file(OUTPUT_FILE, driver="GPKG")
    print(f"💾 Fichier fusionné sauvegardé : {OUTPUT_FILE}")
    
    # Libérer la mémoire
    del fused_gdf, gdf_list
    gc.collect()

except Exception as e:
    print(f"❌ Erreur lors de la fusion : {e}")
```
