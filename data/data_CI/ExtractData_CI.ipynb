```python
import os
import shutil
import pandas as pd

# Base du Drive local ou dossier source
DRIVE_BASE = "/content/drive/MyDrive/GTFS_FINAL/GTFS_CLEAN"
# Dossier local pour CI
WORK_BASE = "/content/drive/MyDrive/GTFS_FINAL/GTFS_CI"

# Villes à traiter
NUM_CITIES = 5

# Créer le dossier de travail
os.makedirs(WORK_BASE, exist_ok=True)

# Lister toutes les villes (sous-dossiers)
all_cities = [d for d in os.listdir(DRIVE_BASE) if os.path.isdir(os.path.join(DRIVE_BASE, d))]
selected_cities = all_cities[:NUM_CITIES]

print(f"CI: Traitement des villes : {selected_cities}")

for city in selected_cities:
    src_dir = os.path.join(DRIVE_BASE, city)
    dest_dir = os.path.join(WORK_BASE, city)
    os.makedirs(dest_dir, exist_ok=True)

    # Copier uniquement les fichiers trips_clean.txt et stop_clean.txt s'ils existent
    for filename in ["trips_clean.txt", "stop_clean.txt"]:
        src_file = os.path.join(src_dir, filename)
        if os.path.exists(src_file):
            shutil.copy2(src_file, os.path.join(dest_dir, filename))
            print(f" - {city}: {filename} copié")
        else:
            print(f" - {city}: {filename} non trouvé")

# Pour stop_times_final.csv (global), on filtre par colonne source_folder
DIR_BASE="/content/drive/MyDrive/GTFS_FINAL"
stop_times_file = os.path.join(DIR_BASE, "stop_times_final.csv")
if os.path.exists(stop_times_file):
    df = pd.read_csv(stop_times_file,low_memory=False)
    df_filtered = df[df["source_folder"].isin(selected_cities)]
    df_filtered_path = os.path.join(WORK_BASE, "stop_times_final_filtered.csv")
    df_filtered.to_csv(df_filtered_path, index=False)
    print(f"stop_times_final.csv filtré et enregistré dans {df_filtered_path}")
else:
    print("stop_times_final.csv non trouvé")
```
